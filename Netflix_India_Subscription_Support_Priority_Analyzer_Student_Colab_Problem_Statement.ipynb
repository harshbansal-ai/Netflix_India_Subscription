## Section 6: Data Validation and Relationship Checks

**Developer handover note:** Complete this section as the Python Developer responsible for repairing the inherited internal tool.

**Datasets to use:**
- `subscriptions.csv`
- `subscribers.csv`
- `plan_catalog.csv`
- `payment_attempts.xlsx`

- `support_tickets.csv`

**How to approach:**
- Check required columns for each file.
- Validate subscription, subscriber, and plan relationships.
- Report duplicate IDs, orphan records, invalid dates, and amount anomalies.

**Expected output format:**
- Validation summary DataFrame with check_name, issue_count, severity, notes.

**Hint:**
- Do not silently delete bad records. Flag them as validation evidence.

```python
print('Columns in payment_attempts_df:')
print(all_data['payment_attempts'].columns)
```

```python
# SECTION 6 — Data validation and relationship checks

validation_results = []
flagged_records = {}

def run_validation_check(check_name, issue_count, severity, notes):
    issue_count = int(issue_count)
    if issue_count > 0:
        final_severity = severity
        final_notes = notes
    else:
        final_severity = "None"
        final_notes = "Passed - no issues found."
    validation_results.append({
        "check_name": check_name,
        "issue_count": issue_count,
        "severity": final_severity,
        "notes": final_notes,
    })

# Work on copies so the raw data stays untouched
table_names = ["subscriptions", "subscribers", "plan_catalog",
               "payment_attempts", "cancellation_feedback", "support_tickets"]
dfs = {}
for t in table_names:
    dfs[t] = all_data[t].copy()

# Print real column names (useful for checking)
for name, df in dfs.items():
    print(name, ":", list(df.columns))

def has_cols(name, cols):
    for c in cols:
        if c not in dfs[name].columns:
            return False
    return True

def find_col(name, keyword):
    for c in dfs[name].columns:
        if keyword in c.lower():
            return c
    return None

# 1) Required columns for each file
required_columns = {
    "subscriptions": ["subscription_id", "subscriber_id", "plan_id",
                      "subscription_status", "subscription_start_date"],
    "subscribers": ["subscriber_id"],
    "plan_catalog": ["plan_id"],
    "payment_attempts": ["subscription_id"],
    "cancellation_feedback": ["subscription_id"],
    "support_tickets": ["subscription_id"],
}
for name, cols in required_columns.items():
    missing = [c for c in cols if c not in dfs[name].columns]
    run_validation_check(name + "_required_columns", len(missing), "High",
                         "Missing required columns in " + name + ": " + str(missing))

# 2) Duplicate IDs
dup_targets = [("subscriptions", "subscription_id"),
               ("subscribers", "subscriber_id"),
               ("plan_catalog", "plan_id")]
for name, col in dup_targets:
    if has_cols(name, [col]):
        df = dfs[name]
        is_dup = df[col].duplicated(keep=False)
        not_null = df[col].notna()
        dups = df[is_dup & not_null]
        key = "duplicate_" + col + "_in_" + name
        flagged_records[key] = dups
        run_validation_check(key, len(dups), "High",
                             "Duplicate " + col + " values found in " + name + ".")

# 3) Orphan / relationship checks
def check_orphans(child, child_col, parent, parent_col, label, severity="Medium"):
    if not has_cols(child, [child_col]) or not has_cols(parent, [parent_col]):
        run_validation_check(label + "_skipped", 1, "Medium",
                             "Skipped: column missing in " + child + " or " + parent + ".")
        return
    has_parent = dfs[child][child_col].isin(dfs[parent][parent_col])
    orphans = dfs[child][~has_parent]
    flagged_records[label] = orphans
    run_validation_check(label, len(orphans), severity,
                         child + " rows whose " + child_col + " has no match in " + parent + ".")

check_orphans("subscriptions", "subscriber_id", "subscribers", "subscriber_id",
              "orphan_subscriber_in_subscriptions")
check_orphans("subscriptions", "plan_id", "plan_catalog", "plan_id",
              "orphan_plan_in_subscriptions")
check_orphans("payment_attempts", "subscription_id", "subscriptions", "subscription_id",
              "orphan_payment_attempts")
check_orphans("cancellation_feedback", "subscription_id", "subscriptions", "subscription_id",
              "orphan_cancellation_feedback")
check_orphans("support_tickets", "subscription_id", "subscriptions", "subscription_id",
              "orphan_support_tickets")

# 4) Invalid dates
for name, df in dfs.items():
    date_cols = [c for c in df.columns if "date" in c.lower()]
    for col in date_cols:
        parsed = pd.to_datetime(df[col], errors="coerce")
        bad = df[col].notna() & parsed.isna()
        invalid = df[bad]
        key = "invalid_date_" + name + "." + col
        flagged_records[key] = invalid
        run_validation_check(key, len(invalid), "High",
                             "Unparseable values in " + name + "." + col + ".")

# 5) Amount anomalies (column names auto-detected)
for name in ["subscriptions", "payment_attempts"]:
    col = find_col(name, "amount")
    if col is None:
        run_validation_check(name + "_amount_column_check", 1, "High",
                             "No column containing 'amount' found in " + name + ".")
        continue
    df = dfs[name]
    numeric = pd.to_numeric(df[col], errors="coerce")
    non_numeric = df[df[col].notna() & numeric.isna()]
    missing = df[df[col].isna()]
    negative = df[numeric < 0]
    flagged_records["non_numeric_" + name + "_" + col] = non_numeric
    flagged_records["negative_" + name + "_" + col] = negative
    run_validation_check(name + "_non_numeric_" + col, len(non_numeric), "High",
                         "Non-numeric values in " + col + ".")
    run_validation_check(name + "_missing_" + col, len(missing), "Medium",
                         "Missing values in " + col + ".")
    run_validation_check(name + "_negative_" + col, len(negative), "High",
                         "Negative values in " + col + ".")

# Final summary
validation_df = pd.DataFrame(validation_results,
                             columns=["check_name", "issue_count", "severity", "notes"])
display(validation_df)
n_issues = (validation_df["issue_count"] > 0).sum()
print("Checks run:", len(validation_df), "| Checks with issues:", n_issues)
```

## Section 8: Payment Failure Classification

**Developer handover note:** Complete this section as the Python Developer responsible for repairing the inherited internal tool.

**Datasets to use:**
- `subscriptions.csv`
- `payment_attempts.xlsx`
- `plan_catalog.csv`

**How to approach:**
- Classify payment failures into business categories.
- Consider gateway failure reason, retry count, settlement status, billing cycle, and payment amount mismatch.
- Create payment risk flags at subscription level.

**Expected output format:**
- Payment risk summary table.
- Subscription-level payment feature table.

**Hint:**
- A failed payment is not always the same issue; classify mandate failure, insufficient funds, gateway failure, and amount mismatch separately.

```python
# SECTION 8 — Payment failure classification
# FIXED: Now classifies payment failures and aggregates features by subscription_id.

payments_df = all_data["payment_attempts"].copy()

# Standardize payment status and failure reasons
payments_df["payment_status_clean"] = payments_df["payment_status"].str.lower().str.strip()
payments_df["failure_reason_clean"] = payments_df["failure_reason"].str.lower().str.strip()
payments_df["failure_reason_clean"] = payments_df["failure_reason_clean"].fillna("no failure")

# Classify payment failure categories
def classify_payment_failure(row):
    if row["payment_status_clean"] == "failed":
        if "insufficient funds" in row["failure_reason_clean"]:
            return "Insufficient Funds"
        elif "bank" in row["failure_reason_clean"] or "gateway" in row["failure_reason_clean"]:
            return "Gateway/Bank Issue"
        elif "mandate" in row["failure_reason_clean"]:
            return "Mandate Failure"
        else:
            return "Other Failure"
    elif row["payment_status_clean"] == "pending":
        return "Pending Payment"
    else:
        return "No Failure"

payments_df["payment_failure_category"] = payments_df.apply(classify_payment_failure, axis=1)

# Aggregate payment features by subscription_id
payment_summary_df = payments_df.groupby("subscription_id").agg(
    total_payment_attempts=("payment_id", "count"),
    failed_payment_attempts=("payment_status_clean", lambda x: (x == "failed").sum()),
    pending_payment_attempts=("payment_status_clean", lambda x: (x == "pending").sum()),
    unique_failure_reasons=("failure_reason_clean", lambda x: x[x != "no failure"].nunique())
).reset_index()

# Add a simple payment risk flag
payment_summary_df["payment_risk_flag"] = np.where(
    payment_summary_df["failed_payment_attempts"] > 0,
    True,
    False
)

print("Payment Summary by Subscription (head):")
display(payment_summary_df.head())
print("Payment Summary by Subscription (shape):", payment_summary_df.shape)
```

```python
# SECTION 10 — Cancellation feedback and complaint linking
# FIXED: Aggregates feedback by subscription_id for merging.

feedback_df = all_data["cancellation_feedback"].copy()

# Standardize sentiment and cancellation_reason
feedback_df["sentiment_clean"] = feedback_df["sentiment_tag"].str.lower().str.strip()
feedback_df["cancellation_reason_clean"] = feedback_df["cancellation_reason"].str.lower().str.strip()
feedback_df["cancellation_reason_clean"] = feedback_df["cancellation_reason_clean"].fillna("no reason provided")

# Aggregate feedback records by subscription_id
feedback_summary_df = feedback_df.groupby("subscription_id").agg(
    feedback_count=("feedback_id", "count"),
    refund_requested_count=("refund_requested", lambda x: (x == "Yes").sum()),
    negative_feedback_count=("sentiment_clean", lambda x: ((x == "negative") | (x == "angry") | (x == "very negative")).sum())
).reset_index()

# Identify primary cancellation reason (most frequent)
primary_reason = feedback_df.groupby(["subscription_id", "cancellation_reason_clean"]).size().reset_index(name="count")
primary_reason = primary_reason.loc[primary_reason.groupby("subscription_id")["count"].idxmax()]
feedback_summary_df = feedback_summary_df.merge(primary_reason[["subscription_id", "cancellation_reason_clean"]],
                                                on="subscription_id", how="left")
feedback_summary_df.rename(columns={"cancellation_reason_clean": "primary_cancellation_reason"}, inplace=True)

print("Feedback Summary by Subscription (head):")
display(feedback_summary_df.head())
print("Feedback Summary by Subscription (shape):")
print(feedback_summary_df.shape)
```

## Section 11: Support Ticket SLA Mapping

**Developer handover note:** Complete this section as the Python Developer responsible for repairing the inherited internal tool.

**Datasets to use:**
- `support_tickets.csv`
- `subscriptions.csv`

**How to approach:**
- Clean ticket status and escalation flags.
- Calculate SLA breach using resolution time and unresolved/escalated signals.
- Aggregate support tickets by subscription.

**Expected output format:**
- Ticket summary table with ticket_count, escalated_ticket_count, sla_breached_ticket_count, avg_resolution_time_hours.

```python
# SECTION 11 — Support ticket mapping
# FIXED: Now aggregates ticket features by subscription_id including SLA and escalation.

tickets_df = all_data["support_tickets"].copy()

# Standardize ticket_status and escalation_flag
tickets_df["ticket_status_clean"] = tickets_df["ticket_status"].str.lower().str.strip()
tickets_df["escalation_flag_clean"] = tickets_df["escalation_flag"].str.lower().str.strip()

# Fill missing resolution_time_hours for calculation (e.g., mark as very long)
# For simplicity, if resolution time is missing, we will not consider it for avg,
# but can still flag as unresolved if status indicates.
tickets_df["resolution_time_hours"] = pd.to_numeric(tickets_df["resolution_time_hours"], errors='coerce')

# Calculate SLA breached (e.g., > 48 hours)
tickets_df["sla_breached"] = tickets_df["resolution_time_hours"] > 48

# Classify escalated tickets
tickets_df["is_escalated"] = (tickets_df["escalation_flag_clean"] == "yes") | \
                             (tickets_df["ticket_status_clean"] == "escalated")

# Classify unresolved tickets (assuming 'open' or 'pending' for simplicity)
tickets_df["is_unresolved"] = (tickets_df["ticket_status_clean"] == "open") | \
                              (tickets_df["ticket_status_clean"] == "pending")

# Aggregate tickets by subscription_id
ticket_summary_df = tickets_df.groupby("subscription_id").agg(
    ticket_count=("ticket_id", "count"),
    escalated_ticket_count=("is_escalated", "sum"),
    sla_breached_ticket_count=("sla_breached", "sum"),
    unresolved_ticket_count=("is_unresolved", "sum"),
    avg_resolution_time_hours=("resolution_time_hours", "mean") # mean will ignore NaNs by default
).reset_index()

# Handle cases where all resolution times are NaN for a group
ticket_summary_df["avg_resolution_time_hours"] = ticket_summary_df["avg_resolution_time_hours"].fillna(0)

print("Ticket Summary by Subscription (head):")
display(ticket_summary_df.head())
print("Ticket Summary by Subscription (shape):", ticket_summary_df.shape)
```

## Section 12: Duplicate and Repeat Issue Detection

**Developer handover note:** Complete this section as the Python Developer responsible for repairing the inherited internal tool.

**Datasets to use:**
- `subscriptions.csv`
- `payment_attempts.xlsx`
- `cancellation_feedback.json`
- `support_tickets.csv`

**How to approach:**
- Detect duplicate subscription IDs and duplicate-looking subscription rows.
- Detect repeated failed payment attempts.
- Detect repeated complaints or tickets for the same subscriber.

**Expected output format:**
- Issue quality summary with duplicate_count, repeated_payment_count, repeated_feedback_count, repeated_ticket_count.

**Hint:**
- Repeated operational signals usually matter more than a single isolated issue.

```python
# SECTION 12 — Duplicate and repeat issue detection
# FIXED: Now detects duplicate IDs and repeated operational patterns and prepares a summary by subscription_id.

# Initialize a DataFrame to store repeated issue counts, indexed by subscriber_id
all_subscribers = pd.DataFrame(subscriptions_df['subscriber_id'].unique(), columns=['subscriber_id'])
all_subscribers.set_index('subscriber_id', inplace=True)
repeat_issue_summary = all_subscribers

# 1. Detect duplicate subscription_id records
duplicate_subscription_records = subscriptions_df[subscriptions_df.duplicated(subset=['subscription_id'], keep=False)]
if not duplicate_subscription_records.empty:
    dup_subs_count = duplicate_subscription_records.groupby('subscriber_id')['subscription_id'].count().rename('duplicate_subscription_records_count')
    repeat_issue_summary = repeat_issue_summary.merge(dup_subs_count, left_index=True, right_index=True, how='left')
else:
    repeat_issue_summary['duplicate_subscription_records_count'] = 0
repeat_issue_summary['duplicate_subscription_records_count'].fillna(0, inplace=True)

# 2. Detect repeated failed payments by subscriber
failed_payments_by_subscriber = payments_df[payments_df['payment_status_clean'] == 'failed'].groupby('subscriber_id').size().rename('failed_payment_count')
# Only include subscribers with more than 1 failed payment
failed_payments_by_subscriber = failed_payments_by_subscriber[failed_payments_by_subscriber > 1]
if not failed_payments_by_subscriber.empty:
    repeat_issue_summary = repeat_issue_summary.merge(failed_payments_by_subscriber, left_index=True, right_index=True, how='left')
else:
    repeat_issue_summary['failed_payment_count'] = 0
repeat_issue_summary['failed_payment_count'].fillna(0, inplace=True)

# 3. Detect repeated feedback by subscriber
repeated_feedback_counts = feedback_df.groupby('subscriber_id').size().rename('feedback_count')
# Only include subscribers with more than 1 feedback entry
repeated_feedback_counts = repeated_feedback_counts[repeated_feedback_counts > 1]
if not repeated_feedback_counts.empty:
    repeat_issue_summary = repeat_issue_summary.merge(repeated_feedback_counts, left_index=True, right_index=True, how='left')
else:
    repeat_issue_summary['feedback_count'] = 0
repeat_issue_summary['feedback_count'].fillna(0, inplace=True)

# 4. Detect repeated support tickets by subscriber
repeated_ticket_counts = tickets_df.groupby('subscriber_id').size().rename('ticket_count')
# Only include subscribers with more than 1 ticket
repeated_ticket_counts = repeated_ticket_counts[repeated_ticket_counts > 1]
if not repeated_ticket_counts.empty:
    repeat_issue_summary = repeat_issue_summary.merge(repeated_ticket_counts, left_index=True, right_index=True, how='left')
else:
    repeat_issue_summary['ticket_count'] = 0
repeat_issue_summary['ticket_count'].fillna(0, inplace=True)


# Filter for subscribers with at least one type of repeated issue
repeat_issue_summary = repeat_issue_summary[
    (repeat_issue_summary['duplicate_subscription_records_count'] > 0) |
    (repeat_issue_summary['failed_payment_count'] > 0) |
    (repeat_issue_summary['feedback_count'] > 0) |
    (repeat_issue_summary['ticket_count'] > 0)
]

print("--- Repeat Issue Summary by Subscriber (counts of issues) ---")
display(repeat_issue_summary.head())
print("Repeat Issue Summary by Subscriber (shape):", repeat_issue_summary.shape)

# Prepare repeat_issue_summary by subscription_id for merging in Section 13
# Merge repeat_issue_summary (by subscriber_id) with subscriptions_df to link subscription_id
subscriptions_with_repeated_issues = subscriptions_df.merge(
    repeat_issue_summary.reset_index(),
    on='subscriber_id',
    how='left'
)

# Fill NaN values for subscribers with no repeated issues
for col in ['duplicate_subscription_records_count', 'failed_payment_count', 'feedback_count', 'ticket_count']:
    subscriptions_with_repeated_issues[col] = subscriptions_with_repeated_issues[col].fillna(0)

# Rename columns to avoid conflicts and provide clearer meaning in final feature table
subscriptions_with_repeated_issues.rename(columns={
    'duplicate_subscription_records_count': 'subscriber_dup_subscription_records_count',
    'failed_payment_count': 'subscriber_repeat_failed_payment_count',
    'feedback_count': 'subscriber_repeat_feedback_count',
    'ticket_count': 'subscriber_repeat_ticket_count'
}, inplace=True)

# For simplicity, carry over subscriber-level issue counts to each of their subscriptions.
# If a subscriber has multiple subscriptions, each will reflect the subscriber's overall issue counts.
repeat_issue_summary_by_subscription_df = subscriptions_with_repeated_issues[
    ['subscription_id', 'subscriber_dup_subscription_records_count', 'subscriber_repeat_failed_payment_count', 'subscriber_repeat_feedback_count', 'subscriber_repeat_ticket_count']
]

print("\nRepeat Issue Summary by Subscription (head, ready for merge):")
display(repeat_issue_summary_by_subscription_df.head())
print("Repeat Issue Summary by Subscription (shape):", repeat_issue_summary_by_subscription_df.shape)
```

## Section 14: Priority Classification

**Developer handover note:** Complete this section as the Python Developer responsible for repairing the inherited internal tool.

**Datasets to use:**
- `final_feature_table`

**How to approach:**
- Assign a numeric support issue score.
- Classify support priority as Critical, High, Medium, or Low.
- Use multiple business signals rather than one column.

**Expected output format:**
- Table with subscription_id, issue_score, support_priority, and supporting signals.

**Hint:**
- Critical cases usually combine failed payment, cancellation/refund request, angry feedback, escalated ticket, and account impact.

```python
# SECTION 14 — Priority classification
# FIXED: Priority is now classified using a comprehensive issue score from multiple signals.

# Define weights for different signals to calculate an overall issue score
# These weights can be tuned based on business priorities
weights = {
    'failed_payment_attempts': 5, # High impact
    'pending_payment_attempts': 2, # Medium impact
    'payment_risk_flag': 3, # General risk
    'negative_feedback_count': 4, # Strong signal of dissatisfaction
    'refund_requested_count': 5, # Direct financial risk
    'escalated_ticket_count': 5, # Requires urgent attention
    'sla_breached_ticket_count': 4, # Indicates operational failure
    'unresolved_ticket_count': 3, # Ongoing issue
    'subscriber_dup_subscription_records_count': 2, # Data quality/churn risk
    'subscriber_repeat_failed_payment_count': 3, # Recurring payment issues
    'subscriber_repeat_feedback_count': 2, # Recurring feedback
    'subscriber_repeat_ticket_count': 2, # Recurring support needs
    'subscription_status_cancelled': 5, # Already lost customer, but might need follow-up
    'subscription_status_expired': 4, # Lost customer
    'subscription_status_grace_period': 3, # At risk
    'subscription_status_payment_hold': 4 # Payment issue
}

# Initialize issue_score
final_feature_table['issue_score'] = 0

# Convert relevant status columns to numeric flags for scoring
final_feature_table['subscription_status_cancelled'] = (final_feature_table['subscription_status'].str.lower() == 'cancelled').astype(int)
final_feature_table['subscription_status_expired'] = (final_feature_table['subscription_status'].str.lower() == 'expired').astype(int)
final_feature_table['subscription_status_grace_period'] = (final_feature_table['subscription_status'].str.lower() == 'grace period').astype(int)
final_feature_table['subscription_status_payment_hold'] = (final_feature_table['subscription_status'].str.lower() == 'payment hold').astype(int)

# Calculate issue_score based on weights and available features
for feature, weight in weights.items():
    if feature in final_feature_table.columns:
        # Handle boolean features (like payment_risk_flag) and numeric counts
        if final_feature_table[feature].dtype == 'bool':
            final_feature_table['issue_score'] += final_feature_table[feature].astype(int) * weight
        else:
            # For counts, we can multiply the count by the weight. Cap at a reasonable max if needed.
            final_feature_table['issue_score'] += np.minimum(final_feature_table[feature], 5) * weight # Cap counts at 5 for score contribution
    else:
        print(f"Warning: Feature '{feature}' not found in final_feature_table. Skipping for issue score calculation.")

# Define priority thresholds (these can also be tuned)
def classify_priority(score):
    if score >= 20:
        return "Critical"
    elif score >= 10:
        return "High"
    elif score >= 3:
        return "Medium"
    else:
        return "Low"

final_feature_table['support_priority'] = final_feature_table['issue_score'].apply(classify_priority)

print("Priority Classification (head):")
display(final_feature_table[["subscription_id", "issue_score", "support_priority",
                               'failed_payment_attempts', 'negative_feedback_count', 'escalated_ticket_count', 'sla_breached_ticket_count', 'subscriber_repeat_ticket_count']].head())
print("Priority distribution:")
display(final_feature_table['support_priority'].value_counts())
```
